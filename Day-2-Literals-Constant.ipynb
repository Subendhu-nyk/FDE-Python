## Literals : A literal values wrtitten directly

```python
# Literals are fixed values written directly in Python code.
```

```python
# 1. Integer Literal
age = 15
# age = Variable | 15 = Integer Literal

# 2. Float Literal
price = 99.99
# price = Variable | 99.99 = Float Literal

# 3. String Literal
name = "FDE"
# name = Variable | "FDE" = String Literal

# 4. Boolean Literal
is_active = True
# is_active = Variable | True = Boolean Literal

# 5. None Literal
result = None
# result = Variable | None = None Literal

# 6. List Literal
numbers = [1, 2, 3]
# numbers = Variable | [1, 2, 3] = List Literal

# 7. Tuple Literal
coordinates = (10, 20)
# coordinates = Variable | (10, 20) = Tuple Literal

# 8. Set Literal
unique_values = {1, 2, 3}
# unique_values = Variable | {1, 2, 3} = Set Literal

# 9. Dictionary Literal
user = {"name": "Veera", "age": 23}
# user = Variable | {"name": "Veera", "age": 23} = Dictionary Literal
```

### Remember:

Mutable = Can be modified.

Immutable = Cannot be modified.

Ordered = Maintains element order.

Unique = Does not allow duplicate elements.

# Constants : unchanged value

### Constants are variables whose values are intended to remain unchanged throughout a program.

### Python does not have a special keyword to declare constants. By convention, we write constant names using UPPERCASE letters.

```python
# Variable (value can be changed)
age = 25
age = 30

print(age)  # Output: 30


# Constants (intended to remain unchanged)
DB_PORT = 3306
PI = 3.14

print(DB_PORT)
print(PI)
```

```python
# Variable
age = 25
age = 30  # Allowed

# Constant
PI = 3.14
PI = 3.14159  # Technically allowed in Python, but avoid changing it

# Python does not enforce constants. Uppercase naming is a convention that tells other developers not to change the value.
```

### Remember
##Variable: A value that can change.

Constant: A value intended to remain unchanged.

Naming convention: Use uppercase letters for constants.
