## Data type

#### A data type classifies a value and determines the kinds of operations that can be performed on it. For example, numbers support arithmetic, strings support text operations, and dictionaries associate keys with values.

#### Python is dynamically typed: we do not normally declare a variable's type in advance. Python determines the type from the object assigned to the variable.


```python
age =25
print(age)
print(type(age))
print(type(age).__name__)
```

```python
# type(age) returns the type object.
# .__name__ gives the readable type name.
# print() displays the returned values.
# The variable name age is not permanently restricted to integers. Reassigning it to another object changes what it refers to.
```

```python
age =25
age="twenty-five"
print(age)
print(type(age).__name__)
```

### 2. Numeric and Boolean data types

### 2.1 Integer (int)
#### An integer represents a whole number without a fractional component. It can be positive, negative or zero. Python integers can grow beyond the fixed-width integer limits found in many other languages, subject to available memory.


```python
age=25
item_count=10
temp=-10
zero_value=0

print(type(age).__name__)
print(type(temp).__name__)
```

```python
# Integer examples include counts, quantities, indexes and many identifiers. 
# An identifier such as a postal code may look numeric but is often better stored as a string if arithmetic is not meaningful and leading zeroes matter.
```

### 2.2 Floating-point (float)

#### A float represents a real-number approximation using binary floating-point storage. It is commonly used for measurements, percentages and values with fractional parts.

##### Important: many decimal fractions cannot be represented exactly in binary floating-point. This is why some calculations display a tiny rounding difference.

```python
price = 25.0
temperature = -2.5

print(type(price).__name__)
print(0.1 + 0.2)
```

```python
# The small difference is a floating-point representation effect, not a Python arithmetic bug. 
# For financial calculations requiring exact decimal arithmetic
```

```python
from decimal import Decimal

total = Decimal("0.1") + Decimal("0.2")
print(total)
```

### 2.3 Complex (complex)

#### A complex number has a real part and an imaginary part. Python uses j to represent the imaginary unit.

#### A complex value has the form a + bj, where a is the real component and b is the imaginary coefficient.

```python
signal_value = 2 + 1j

print(signal_value)
print(signal_value.real)
print(signal_value.imag)
print(type(signal_value).__name__)
```

### 2.4 Boolean (bool)

#### A Boolean represents a logical truth value: True or False. The capital letters matter.

#### Booleans are used in conditions, filters, validation and decisions. In Python, bool is a subclass of int, so True behaves numerically like 1 and False like 0; this is a language detail, not a reason to use them as ordinary counts.

```python
is_active = True
is_completed = False

print(type(is_active).__name__)
print(is_active)
print(True + True)
```

```python

# Python allows the + operator to work with Boolean values because Python's bool type is a subclass of int. 
# Therefore, Boolean values can participate in mathematical operations.
# Python does not permanently convert True into 1.
# When used in arithmetic, Python treats True as 1 and False as 0.
# The + operator performs addition, not a Boolean operation.
```

### 3. String (str)

#### A string is an immutable sequence of Unicode characters used to represent text. Strings can be written using single quotes, double quotes or triple quotes.

#### A numeric-looking value inside quotes is still text. For example, "45" is a string, while 45 is an integer.



```python
course = "Python"
code_text = 'AI-101'
description = """Python is used in data science and AI."""

print(type(course).__name__)
print(type("45").__name__)
print(type(45).__name__)
print(description)
```

#### 3.1 String indexing
##### Indexing retrieves one character. Python uses zero-based indexing: the first character is at index 0. Negative indexes count from the end.

```python
language = "Python"

print(language[0])
print(language[1])
print(language[-1])
```

#### 3.2 String slicing

##### Slicing extracts a portion of a sequence using sequence[start:stop:step]. The start position is included; the stop position is excluded.


```python
language = "Python"

print(language[0:3])
print(language[2:])
print(language[:4])
print(language[::2])
```

### 4. Collection data types

#### A collection stores multiple values in one object. Python's main built-in collection types differ in ordering, mutability, uniqueness and access patterns.

#### 4.1 List (list)
##### A list is an ordered, mutable collection. It can contain duplicate values and values of different types.

##### Lists are useful when a collection may change, such as a list of records, observations or model predictions.


```python
scores = [85, 90, 78, 90]
print(scores)
print(type(scores).__name__)
print(scores[0])
print(scores[-1])
```

```python
# Modify a list
# Because lists are mutable, we can add, update and remove elements.

scores = [85, 90, 78]

scores.append(95)
print(scores)
scores[0] = 88
print(scores)
```

#### 4.2 Tuple (tuple)
#### A tuple is an ordered, immutable collection. It allows duplicates and can contain different data types.

#### Tuples are useful for fixed groups of related values, such as coordinates or a function returning multiple results.

```python
location = (20.2961, 85.8245)
student_record = ("Ravi", 25, "FDE")

print(location)
print(student_record[0])
print(type(student_record).__name__)
```

```python
# Tuple unpacking
# Unpacking assigns tuple elements to separate variables. The number of variables must normally match the number of elements.

student_name, student_age, course_name = student_record

print(student_name)
print(student_age)
print(course_name)
# A tuple itself cannot have an element reassigned. However, if it contains a mutable object such as a list, that nested object may still be modified. 
# Immutability is not automatically deep.
```

#### 4.3 Range (range)
##### A range represents an immutable arithmetic sequence of integers. It is memory-efficient for iteration because it does not need to store every integer as a separate list element.

##### Syntax: range(start, stop, step)
##### - start defaults to 0.
##### - stop is excluded.
##### - step defaults to 1 and cannot be zero.

```python
numbers = range(5)

print(numbers)
print(list(numbers))
print(type(numbers).__name__)
```

```python
print(list(range(2, 10, 2)))
print(list(range(5, 0, -1)))
```

#### 4.4 Set (set)
##### A set is a mutable collection of distinct, hashable elements. Duplicate values are removed. Sets are useful for membership tests and mathematical set operations.

##### An empty set is written as set(), not {}. The expression {} creates an empty dictionary.

```python
unique_numbers = {10, 20, 30, 30}

print(unique_numbers)
print(type(unique_numbers).__name__)
print(len(unique_numbers))
```

```python
empty_set = set()
empty_dictionary = {}

print(type(empty_set).__name__)
print(type(empty_dictionary).__name__)
```

#### 4.5 Dictionary (dict)
#### A dictionary stores key-value associations. Each key must be unique and hashable; values can be repeated and can be of any type.

#### Dictionaries are useful for structured records, configuration, lookup tables and JSON-like data.

```python
student = {
    "name": "Ravi",
    "age": 25,
    "course": "FDE"
}

print(student)
print(type(student).__name__)
print(student["name"])
```

### 5. None and the NoneType
#### None is Python's single value representing the absence of a value or an intentionally empty result. Its type is NoneType.

#### It is different from 0, False, "", [] and {}. Those are actual values, even though some are considered falsy in conditions.

```python
result = None

print(result)
print(type(result).__name__)
print(result is None)
```
